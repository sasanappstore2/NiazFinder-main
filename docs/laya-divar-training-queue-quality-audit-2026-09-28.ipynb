# ممیزی کیفیت داده و صف آموزش Laya — ۲۸ سپتامبر ۲۰۲۶

این یادداشت وضعیت قابل‌بازتولید corpus و صف مرحله‌ای آموزش را ثبت می‌کند. منبع مجاز فقط Divar است. ردیف‌های Divar آگهی عرضهٔ فروشنده/مشاورند، نه نیاز ثبت‌شدهٔ متقاضی. هر متن نیاز در این مسیر فرضی/مشتق‌شده است؛ هیچ‌کدام ground truth واقعی یا تأیید انسانی نیست.

## یافته‌های snapshot

- از ۱٬۰۰۰٬۰۰۰ ردیف منبع، ۹۹۹٬۴۱۶ آگهی retained مانده و ۵۸۴ ردیف به‌علت تعارض/دسته حذف شده‌اند. آمار builder تعداد ۹۹۶٬۱۱۱ گروه متن نرمال‌شده را پیش از فیلتر تعارض ثبت می‌کند؛ در facts retained ممیزی مستقل ۳٬۳۴۱ گروه تکراری و ۳٬۵۴۱ ردیف مازادِ متن را یافت که همگی داخل split خود ماندند.
- فقط ۱۶ مورد از ۱۸ برگ دستهٔ ملکی برنامه در Divar پوشش دارند؛ `agency-services` و `land-rent` غایب‌اند.
- شهر: ۳۹۷ از ۴۲۱ شهر منبع به کاتالوگ برنامه نگاشت شده‌اند (۹۴٫۳۰٪)، اما این فقط ۳۲٫۸۹٪ از ۱٬۲۰۷ کاتالوگ شهر برنامه است. ۹۶٫۷۷٪ آگهی‌های نگه‌داشته‌شده شهر نگاشت‌شده دارند.
- محله: crosswalk دقیق v4 تعداد ۱٬۱۵۸ از ۱٬۱۸۸ زوج شهر-محله را resolve می‌کند (۹۷٫۴۷٪)؛ ۳۰ زوج به‌دلیل نبود شناسهٔ رسمی در API فعلی unresolved مانده‌اند. بازسازی facts v5 نشان داد ۷۴۸ ردیف retained جدید با تطبیق لفظیِ یکتا به کاتالوگ همان شهر نگاشت شدند؛ پوشش محله از ۴۳۰٬۲۶۸ به ۴۳۱٬۰۱۶ ردیف (۴۳٫۱۳٪) رسید. هیچ فیلدی جز `offerLocation` تغییر نکرد.
- ممیزی مستقل v5: هر ۹۹۹٬۴۱۶ ردیف خروجی و ورودی با هم parse و مقایسه شد؛ hash هر دو سمت با زنجیرهٔ manifest منطبق است؛ ۷۴۸ تغییر مکانی مشاهده شد، ۰ فیلد غیرمکانی و ۰ فیلد مکانی خارج از allowlist تغییر کرد. JSON خراب، شناسه/ordinal تکراری، split leakage، شهر/شناسهٔ محلهٔ نامعتبر و تماس باقی‌مانده در scan الگوها صفر بود. ۳٬۵۴۱ ردیف تکرار متن درون همان split حفظ شده‌اند؛ بنابراین distinct-example count را نباید با تعداد ردیف برابر گرفت. این داده همچنان آگهی عرضه است، نه نیاز واقعی؛ حقوق استفاده و بازبینی جامع حریم خصوصی باز هستند.
- corpusهای v3 و v6 فرضی، بدون بازبینی انسانی و صراحتاً training-ineligible هستند. متن خروجی Laya نیز تا وقتی مستقل ارزیابی/بازبینی نشده، برچسب طلایی نیست.
- دسته‌ها نامتوازن‌اند؛ در v6 تعداد `apartment-sale` برابر ۲۲۱٬۱۱۷ و `workspace-short-rent` برابر ۵۳۴ است. trainer بعدی RLCD اکنون وزن‌دهی معکوس ریشه‌ای per-field با سقف ۵ برابر را در objective، preflight و manifest ثبت می‌کند؛ این وزن‌دهی نامتوازنی باقیمانده را حذف نمی‌کند.
- صف‌ها سریالی‌اند: v3 جاری → v6 آموزش RLCD → v7 ارزیابی کامل → v8 تبدیل محلی با همان Laya. هیچ مرحلهٔ تازه‌ای نباید تا آزادشدن MPS و عبور از gate قبلی شروع شود.
- پایش زندهٔ ۲۳:۲۶ به وقت تهران: PID 56641 همچنان زنده و روی `--device mps` بود؛ نمونهٔ stack مسیر PyTorch→Metal/MPSGraph را نشان داد. آخرین checkpoint در ۲۳:۰۷ روی step=27٬۰۰۰/54٬۲۵۸ ثبت شده بود؛ حدود ۱۹ دقیقه از آن گذشته و هنوز زیر مهلت stale سه‌ساعته بود. v6 منتظر خروج v3، v7 منتظر gate v6 و v8 منتظر benchmark v7 است؛ MPS دوم شروع نشده است.
- ارزیابی روش: Laya تصمیم typed در یک forward pass است و متن نیاز تولید نمی‌کند. طبق راهنمای رسمی، calibration باید روی دادهٔ جدا از train سنجیده شود؛ checkpoint چندزبانه نیز برای score سوگیری ترتیب گزینه دارد، پس انتخاب محله/دسته با `choice` و ارزیابی held-out انجام می‌شود، نه تبدیل score یا confidence خام به حقیقت. [راهنمای رسمی Fine-Tuning](https://github.com/NandhaKishorM/laya#fine-tune-for-better-accuracy) و [Model card مدل چندزبانه](https://huggingface.co/convaiinnovations/laya-multilingual).

انتقال ابری خاموش است؛ مجوز downstream/بازنشر و بازبینی جامع حریم خصوصی همچنان باز هستند. این ممیزی به‌تنهایی مجوز بازنشر یا استفادهٔ تولیدی نیست.

```python
from pathlib import Path
import json
import subprocess
import time

ROOT = Path.cwd()
def load_json(path):
    return json.loads((ROOT / path).read_text(encoding='utf-8'))

coverage_run = subprocess.run(
    ['bun', 'scripts/datasets/audit-divar-app-coverage.ts'],
    cwd=ROOT, check=True, capture_output=True, text=True,
)
coverage = json.loads(coverage_run.stdout)
v3 = load_json('data/divar/divar-hypothetical-needs-laya-v3-exact-city-aliases-2026-09-28.jsonl.manifest.json')
v6 = load_json('data/divar/divar-hypothetical-needs-v6-explicit-rent-mode-2026-09-28.jsonl.manifest.json')
progress = load_json('data/laya-experiments/divar-hypothetical-needs-v3-mps-full-2026-09-28/training-progress.json')
progress_path = ROOT / 'data/laya-experiments/divar-hypothetical-needs-v3-mps-full-2026-09-28/training-progress.json'
literal_crosswalk = load_json('data/divar/official-neighborhood-app-crosswalk-2026-09-28-literal-exact-v4.json')
v5_audit = load_json('data/divar/divar-property-offer-facts-v5-literal-neighborhood-crosswalk-row-quality-lineage-audit-2026-09-28-r2.json')
plans = {
    'v6': load_json('data/laya-experiments/divar-v6-explicit-rent-mode-mps-full-2026-09-28.launch-plan.json'),
    'v7': load_json('data/laya-experiments/divar-v7-neighborhood-benchmark-after-v6-2026-09-28.json'),
    'v8_shadow': load_json('data/laya-experiments/divar-v8-laya-shadow-after-v7-2026-09-28.json'),
}
categories = v6['categoryCounts']
summary = {
    'v4_baseline_coverage': coverage,
    'v3': {key: v3.get(key) for key in ('outputRows', 'synthetic', 'realNeedGroundTruth', 'humanReviewed', 'trainingEligible', 'layaPredictionsUsedAsLabels')},
    'v6': {key: v6.get(key) for key in ('outputRows', 'synthetic', 'realNeedGroundTruth', 'humanReviewed', 'trainingEligible', 'layaPredictionsUsedAsLabels')},
    'v6_category_skew_max_to_min': max(categories.values()) / min(categories.values()),
    'active_v3_progress': progress,
    'active_v3_progress_age_seconds': round(time.time() - progress_path.stat().st_mtime),
    'literal_exact_crosswalk': {
        'coverage': literal_crosswalk['coverage'],
        'literal_exact_pairs': sum(entry.get('matchBasis') == 'unique-literal-official-name' for entry in literal_crosswalk['entries']),
        'unresolved_reasons': literal_crosswalk['coverage'].get('unresolvedReasons', {}),
    },
    'v5_reconciled_facts_audit': {
        'status': v5_audit['status'],
        'rows': v5_audit['rows'],
        'neighborhoodCoverage': {key: v5_audit['referenceCoverage'][key] for key in ('mappedNeighborhoodRows', 'shareOfParsedRowsWithMappedNeighborhoodPct', 'distinctMappedAppCityNeighborhoodPairs')},
        'qualityGates': v5_audit['qualityGates'],
        'lineage': v5_audit['lineage'],
    },
    'queued_stage_statuses': {name: plan['status'] for name, plan in plans.items()},
    'v8_training_eligibility': plans['v8_shadow'].get('limits', []),
}
assert v3['synthetic'] and v3['realNeedGroundTruth'] is False and v3['trainingEligible'] is False
assert v6['synthetic'] and v6['realNeedGroundTruth'] is False and v6['trainingEligible'] is False
assert coverage['transferAndRights']['cloudTransferAllowed'] is False
assert summary['literal_exact_crosswalk']['literal_exact_pairs'] == 6
assert literal_crosswalk['coverage']['sourceRowsWithLiteralExactName'] == 748
assert v5_audit['status'] == 'complete'
assert v5_audit['rows']['rowsParsed'] == 999416 and v5_audit['rows']['malformedJsonRows'] == 0
assert v5_audit['lineage']['newlyMappedRowsObserved'] == 748 and v5_audit['lineage']['nonLocationChangedRowsObserved'] == 0
assert v5_audit['referenceCoverage']['mappedNeighborhoodRows'] == 431016
assert v5_audit['qualityGates']['inputSha256MatchesManifest'] is True
assert v5_audit['qualityGates']['sourceFactsHashMatchesReconciliationManifest'] is True
assert v5_audit['qualityGates']['nonLocationFieldsUnchangedFromSource'] is True
assert v5_audit['qualityGates']['onlyAllowedLocationFieldsChanged'] is True
assert v5_audit['qualityGates']['locationChangedRowsMatchManifest'] is True
print(json.dumps(summary, ensure_ascii=False, indent=2))
```
